1. **What are the key components of a typical machine learning project pipeline?**
   Data collection, data preprocessing, feature engineering, model training, evaluation, and deployment are the main stages of a machine learning pipeline.

2. **Explain the difference between supervised and unsupervised machine learning algorithms.**
   Supervised learning uses labeled data to make predictions, while unsupervised learning finds patterns or groups in unlabeled data.

3. **How do you handle missing values in a machine learning dataset?**
   Missing values can be handled by removing records or filling them using mean, median, mode, or other imputation techniques.

4. **What is cross-validation in machine learning and why is it important?**
   Cross-validation splits data into multiple subsets to test model performance. It helps ensure the model generalizes well to unseen data.

5. **Discuss the bias-variance tradeoff in machine learning models.**
   High bias leads to underfitting, while high variance leads to overfitting. A good model balances both for optimal performance.

6. **Explain the concept of feature engineering and its importance in machine learning.**
   Feature engineering involves creating or modifying features to improve model learning and predictive accuracy.

7. **What is the purpose of hyperparameter tuning in machine learning models?**
   Hyperparameter tuning finds the best model settings to improve performance and reduce prediction errors.

8. **How does the K-means clustering algorithm work? What are its limitations?**
   K-means groups data into K clusters by assigning points to the nearest centroid. It requires choosing K beforehand and is sensitive to outliers.

9. **Describe the difference between overfitting and underfitting in machine learning models.**
   Overfitting occurs when a model learns noise from training data, while underfitting occurs when the model is too simple to capture patterns.

10. **What is the ROC curve and how is it used to evaluate machine learning models?**
    The ROC curve plots the true positive rate against the false positive rate. It helps evaluate a classification model's performance.

11. **Explain the importance of regularization techniques in preventing overfitting.**
    Regularization adds penalties to complex models, reducing overfitting and improving generalization on new data.

12. **What is ensemble learning and how does it improve predictive performance?**
    Ensemble learning combines multiple models to make predictions, often achieving higher accuracy and robustness than a single model.

13. **How can you assess the performance of a machine learning model using evaluation metrics?**
    Metrics such as Accuracy, Precision, Recall, F1-score, MSE, and R² are used to measure model effectiveness depending on the task.

14. **Discuss the concept of transfer learning in the context of deep learning models.**
    Transfer learning uses a pre-trained model as a starting point for a new task, reducing training time and improving performance with limited data.
