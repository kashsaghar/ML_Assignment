### 1. What is the purpose of model evaluation in machine learning projects?

Model evaluation measures how well a trained model performs on unseen data and helps select the best model.

---

### 2. Explain the difference between training error and test error in model evaluation.

* **Training error:** Error on the training dataset.
* **Test error:** Error on unseen test data, showing the model's ability to generalize.

---

### 3. What are some common metrics used for evaluating classification models?

Common metrics include Accuracy, Precision, Recall, F1-score, ROC-AUC, and Confusion Matrix.

---

### 4. How does cross-validation help in improving model evaluation?

Cross-validation evaluates the model on different data splits, providing a more reliable estimate of performance.

---

### 5. Explain the concept of overfitting in the context of model evaluation.

Overfitting occurs when a model learns the training data too well, including noise, and performs poorly on new data.

---

### 6. Why is it important to split the data into training and testing sets during model evaluation?

It allows the model to be trained on one dataset and evaluated on unseen data to measure its real performance.

---

### 7. What is a confusion matrix and how is it used in evaluating classification models?

A confusion matrix shows the number of correct and incorrect predictions, helping calculate metrics like precision and recall.

---

### 8. Discuss the bias-variance tradeoff in model evaluation.

A good model balances bias and variance to achieve high accuracy on both training and test data.

---

### 9. What is precision and recall in the context of model evaluation?

* **Precision:** Percentage of predicted positive cases that are actually positive.
* **Recall:** Percentage of actual positive cases correctly identified.

---

### 10. Explain the purpose of ROC curve and AUC in evaluating classification models.

The ROC curve shows the trade-off between true positive rate and false positive rate. AUC measures the model's overall classification performance.

---

### 11. How can you handle imbalanced classes in model evaluation?

Use techniques like SMOTE, oversampling, undersampling, class weighting, and metrics such as Precision, Recall, and F1-score.

---

### 12. What is the F1-score and why is it useful in model evaluation?

The F1-score is the harmonic mean of precision and recall. It is useful for evaluating models on imbalanced datasets.

---

### 13. Discuss the role of hyperparameter tuning in improving model evaluation.

Hyperparameter tuning finds the best parameter values to improve model accuracy and overall performance.
