### 1. What is data ingestion in a machine learning project?

Data ingestion is the process of collecting and importing data from various sources into a system for storage, processing, and analysis.

### 2. Why is data ingestion important?

It ensures that accurate, relevant, and up-to-date data is available for training and evaluating machine learning models.

### 3. Common sources of data for machine learning projects?

* Databases (SQL, NoSQL)
* APIs
* CSV/Excel files
* Sensors and IoT devices
* Web scraping
* Cloud storage

### 4. How is data retrieved from a relational database?

Using SQL queries to extract required data from tables, then loading it into tools like Python Pandas for analysis and modeling.

### 5. Challenges during data ingestion?

* Missing or incomplete data
* Inconsistent formats
* Large data volumes
* Data duplication
* Security and privacy concerns

### 6. Role of data preprocessing in data ingestion?

Preprocessing cleans and transforms raw data by handling missing values, removing duplicates, and standardizing formats.

### 7. How can data ingestion be automated?

Using ETL pipelines, scripts, schedulers (e.g., Cron), or workflow tools like Apache Airflow.

### 8. Batch processing vs Real-time processing

* **Batch Processing:** Data is collected and processed at scheduled intervals.
* **Real-Time Processing:** Data is processed immediately as it arrives.

### 9. What is data enrichment?

Adding extra information from external or internal sources to improve the quality and usefulness of data.

### 10. Role of data governance?

It establishes policies and standards to ensure data quality, security, compliance, and consistency.

### 11. How can data lineage be maintained?

By tracking data sources, transformations, and movement through logs, metadata, and documentation.

### 12. Importance of data sampling techniques?

Sampling reduces dataset size while maintaining representativeness, making analysis faster and more efficient.

### 13. What is data wrangling?

Data wrangling is the process of cleaning, organizing, and transforming raw data into a usable format for analysis.

### 14. Best practices for data consistency and integrity?

* Validate data before loading
* Remove duplicates
* Standardize formats
* Use automated quality checks
* Maintain proper documentation and version control
